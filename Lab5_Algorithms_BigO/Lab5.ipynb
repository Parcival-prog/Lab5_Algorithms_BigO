```python
# EMERGENCY DISPATCH RECORD SEARCH
ID_SUM, SEED, ARTIST_SUM = 25, 7, 98
N = 15

original = [ID_SUM + SEED * i for i in range(N)]
rotated = original[SEED:] + original[:SEED]
target = ID_SUM + ARTIST_SUM   # 123

def linear_search(a, t):
    comps = 0
    for i, v in enumerate(a):
        comps += 1
        if v == t:
            return i, comps
    return -1, comps

def rotated_binary_search(a, t):
    lo, hi, comps = 0, len(a) - 1, 0
    while lo <= hi:
        mid = (lo + hi) // 2
        comps += 1
        if a[mid] == t:
            return mid, comps
        if a[lo] <= a[mid]:              #
            if a[lo] <= t < a[mid]:
                hi = mid - 1
            else:
                lo = mid + 1
        else:                            
            if a[mid] < t <= a[hi]:
                lo = mid + 1
            else:
                hi = mid - 1
    return -1, comps

print("Original:", original)
print("Rotated: ", rotated)
print("Target:", target)
for name, f in [("Linear", linear_search), ("Binary", rotated_binary_search)]:
    pos, c = f(rotated, target)
    print(name, "-> index", pos, "| comparisons:", c)
```

```python
# UNIVERSITY ENROLLMENT RECORDS
ID_SUM, SEED, ARTIST_LEN, NAME_LEN = 25, 7, 10, 7

A = [ID_SUM + SEED * i for i in range(8)]
B = [ARTIST_LEN + NAME_LEN * i for i in range(7)]

def merge(a, b):
    i = j = comps = 0
    out = []
    while i < len(a) and j < len(b):
        comps += 1
        if a[i] <= b[j]:
            out.append(a[i]); i += 1
        else:
            out.append(b[j]); j += 1
    out += a[i:] + b[j:]
    return out, comps

merged, comps = merge(A, B)

print("A:", A)
print("B:", B)
print("Merged:", merged)
print("Records before:", len(A), "+", len(B), "=", len(A) + len(B))
print("Records after:", len(merged))
print("Comparisons:", comps)
print("Sorted?", merged == sorted(merged))
```

```python
# E-COMMERCE ORDER PRCESSING
import random, time

ID_SUM, SEED, N = 25, 7, 15
orders = [ID_SUM + SEED*i for i in range(N)]
random.seed(SEED)
rnd = orders[:]; random.shuffle(rnd)
conds = {"Sorted": orders[:], "Reverse": orders[::-1], "Random": rnd}

def bubble(a):
    a=a[:]; c=m=0
    for i in range(len(a)-1):
        swapped=False
        for j in range(len(a)-1-i):
            c+=1
            if a[j]>a[j+1]:
                a[j],a[j+1]=a[j+1],a[j]; m+=1; swapped=True
        if not swapped: break
    return a,c,m

def selection(a):
    a=a[:]; c=m=0
    for i in range(len(a)-1):
        k=i
        for j in range(i+1,len(a)):
            c+=1
            if a[j]<a[k]: k=j
        if k!=i: a[i],a[k]=a[k],a[i]; m+=1
    return a,c,m

def insertion(a):
    a=a[:]; c=m=0
    for i in range(1,len(a)):
        key=a[i]; j=i-1
        while j>=0:
            c+=1
            if a[j]>key: a[j+1]=a[j]; m+=1; j-=1
            else: break
        a[j+1]=key
    return a,c,m

# ---- Datasets ----
print("Generated order dataset:", orders)
print("Sorted input: ", conds["Sorted"])
print("Reverse input:", conds["Reverse"])
print("Random input: ", conds["Random"])
print()

# ---- Results ----
final_sets = {}
for cn, d in conds.items():
    for n, f in [("Bubble", bubble), ("Selection", selection), ("Insertion", insertion)]:
        t = time.perf_counter()
        for _ in range(1000):
            r, c, m = f(d)
        us = (time.perf_counter() - t) / 1000 * 1e6
        final_sets[(cn, n)] = r
        print(f"{cn:8} {n:10} comparisons: {c:3} | moves: {m:3} | "
              f"time: {us:6.1f} µs | sorted: {r == sorted(d)}")

# ---- Final organized set ----
print()
print("Final organized (sorted) set:")
for (cn, n), r in final_sets.items():
    print(f"  {cn:8} + {n:10}: {r}")

print()
print("All results identical and correct:",
      all(r == orders for r in final_sets.values()))
```
